### 3. Canonical City Normalization

```python
def normalize_city(city):
    """
    Convert city names to one canonical representation.

    Examples
    --------
    "Bucharest"     -> "bucharest"
    " Iași "        -> "iasi"
    "Iasi"          -> "iasi"
    "Timișoara"     -> "timisoara"

    "Cluj-Napoca"   -> "cluj_napoca"
    "Cluj_Napoca"   -> "cluj_napoca"
    "Cluj Napoca"   -> "cluj_napoca"
    "Cluj–Napoca"   -> "cluj_napoca"

    Do not modify this function.
    """

    city = str(city).strip().lower()

    # Remove accents / diacritics.
    city = unicodedata.normalize(
        "NFKD",
        city,
    )

    city = "".join(
        character
        for character in city
        if not unicodedata.combining(character)
    )

    # Convert spaces, underscores and different dash
    # characters into one canonical separator.
    city = re.sub(
        r"[\s_\-–—−]+",
        "_",
        city,
    )

    city = city.strip("_")

    return city
```

### 13. Search Node

```python
@dataclass
class Node:

    state: str

    parent: "Node | None" = None

    path_cost: float = 0.0

    depth: int = 0
```

### 14. Successors

```python
def successors(state):
    """
    Return successor states in alphabetical order.

    Do not modify this function.
    """

    state = normalize_city(
        state
    )

    if state not in graph:
        raise KeyError(
            f"Unknown city: {state}"
        )

    return sorted(
        graph[state].keys()
    )
```

### 15. Reconstruct Path

```python
def reconstruct_path(node):

    path = []

    while node is not None:

        path.append(
            node.state
        )

        node = (
            node.parent
        )

    return list(
        reversed(path)
    )
```

### 16. Path Metrics

```python
def path_metrics(path):

    if path is None:
        return {
            "path_length": None,
            "distance_km": None,
            "time_min": None,
        }

    distance = 0.0
    time = 0.0

    for city_a, city_b in zip(
        path[:-1],
        path[1:],
    ):

        distance += (
            graph[city_a]
            [city_b]
            ["distance_km"]
        )

        time += (
            graph[city_a]
            [city_b]
            ["time_min"]
        )

    return {
        "path_length":
            len(path) - 1,

        "distance_km":
            float(distance),

        "time_min":
            float(time),
    }
```

### 17. Validate Path

```python
def validate_path(
    path,
    start,
    goal,
):

    if not path:
        return False

    start = normalize_city(
        start
    )

    goal = normalize_city(
        goal
    )

    # Require canonical city identifiers.
    if path[0] != start:
        return False

    if path[-1] != goal:
        return False

    for city in path:

        if city not in graph:
            return False

    for city_a, city_b in zip(
        path[:-1],
        path[1:],
    ):

        if city_b not in graph[city_a]:
            return False

    return True
```

### 18. Standard Result Format

```python
def build_result(
    algorithm,
    node,
    expanded_nodes,
    generated_nodes,
    max_frontier_size,
    expanded_order,
):

    # --------------------------------------------------------
    # Failure
    # --------------------------------------------------------

    if node is None:

        return {

            "algorithm":
                algorithm,

            "solution": {

                "found":
                    False,

                "path":
                    None,

                "path_length":
                    None,

                "distance_km":
                    None,

                "time_min":
                    None,
            },

            "diagnostics": {

                "expanded_nodes":
                    int(expanded_nodes),

                "generated_nodes":
                    int(generated_nodes),

                "max_frontier_size":
                    int(max_frontier_size),

                "expanded_order":
                    list(expanded_order),
            },
        }

    # --------------------------------------------------------
    # Success
    # --------------------------------------------------------

    path = reconstruct_path(
        node
    )

    metrics = path_metrics(
        path
    )

    return {

        "algorithm":
            algorithm,

        "solution": {

            "found":
                True,

            "path":
                path,

            **metrics,
        },

        "diagnostics": {

            "expanded_nodes":
                int(expanded_nodes),

            "generated_nodes":
                int(generated_nodes),

            "max_frontier_size":
                int(max_frontier_size),

            "expanded_order":
                list(expanded_order),
        },
    }
```

### 19. Metric Definitions

#### expanded_nodes:
Number of nodes removed from the frontier for which successors are generated.

A goal node is not counted as expanded if the search stops immediately after selecting it.


#### generated_nodes:
Number of child search nodes actually created and added to the frontier.
The initial search node is not included.
A successor that is examined but discarded because its state has already been reached (BFS/DFS), or because it does not improve the best known path cost (UCS/A*), is not counted as generated.
#### max_frontier_size:
Maximum number of search nodes simultaneously stored in the frontier.

#### path_length:
Number of actions / graph edges in the returned path.

#### Task 1 - Breadth-First Search

```python
def bfs(graph: dict, start: str, end: str) -> str:

    if start not in graph:
        return []

    if end not in graph:
        return []

    queue = deque()
    visited = set()
    parent = {}



    queue.append(start)
    visited.add(start)
    parent[start] = None


    while queue:

        current = queue.popleft()


        if current == end:
            break

        for city in graph[current]:
            if city not in visited:
                visited.add(city)
                parent[city] = current
                queue.append(city)

    if end not in visited:
        return []

    route = []
    current = end

    while current:
        route = [current] + route
        current = parent[current]

    return route
```

#### Task 2 - Depth-First Search

```python

def dfs(graph: dict, visited: set, current: str,  end: str):

    if current == end:
        return [current]

    visited.add(current)


    for neighbour in graph[current]:
        if neighbour not in visited:
            path = dfs(graph, visited, neighbour,  end)

            if path:
                return [current] + path

    return None

```

#### Task 3 - Uniform-Cost Search

```python
def ucs(graph:dict, start:str, end:str) -> list:
    priority_queue = [(0,start, [start])]
    visited = set()


    while priority_queue:
        cost, current, path = heapq.heappop(priority_queue)

        if current == end:
            return path, cost


        if current in visited:
            continue

        visited.add(current)


        for neighbour in graph[current]:
            new_cost = cost + graph[current][neighbour]["distance_km"]

            if neighbour not in visited:
                heapq.heappush(priority_queue, (new_cost,neighbour,path + [neighbour]))
```

#### Task 4
##### Geographic Heuristic Formula
##### - Haversine distance

lat1 = $lat1_{degrees} * \frac{\pi}{180}$

lat2 = $lat2_{degrees} * \frac{\pi}{180}$

lon1 = $lon1_{degrees} * \frac{\pi}{180}$

lon2 = $lon2_{degrees} * \frac{\pi}{180}$

$\Delta lat = lat_2 - lat_1$

$\Delta lon = lon_2 - lon_1$

a = $\sin^2(\frac{\Delta lat}{2})+\cos(lat1)\cos(lat2)\sin^2(\frac{\Delta lon}{2})$

c = $\arcsin(\sqrt(a))$

distance = $R * c$

R = 6371 km

Angles must be converted from degrees to radians

#### Implementation

```python
def haversine(lat1_degrees: float, long1_degrees: float, lat2_degrees: float, long2_degrees:float):
    lat1 = lat1_degrees * math.pi / 180
    lat2 = lat2_degrees * math.pi / 180
    long1 = long1_degrees * math.pi / 180
    long2 = long2_degrees * math.pi / 180

    delta_lat = lat2 - lat1
    delta_long = long2 - long1

    a = math.sin(delta_lat/2)**2 + math.cos(lat1) * math.cos(lat2) * math.sin(delta_long/2)**2
    c = 2 * math.asin(math.sqrt(a))

    distance = EARTH_RADIUS_KM * c
    return distance
```

#### Distance Heruistic

```python
def h_distance(positions: dict, origin: str, destination: str):
    lat1, lon1 = positions[origin]
    lat2, lon2 = positions[destination]

    return haversine(
        lat1,
        lon1,
        lat2,
        lon2
    )


print(h_distance(positions,"budapest", "rome"))
```

#### Task 5 - Time Heuristic

##### Idea

Straight-line distance is measured in kilometres.

 A lower bound on travel time can be obtained by:

    straight-line distance
    ----------------------
    maximum network speed

##### Implementation

```python
def max_network_speed(graph):
    speeds = []
    for neighbours in graph.values():
        for edge in neighbours.values():
            speeds.append(edge["distance_km"]/edge["time_min"])

    return max(speeds)

MAX_SPEED = max_network_speed(graph)

def h_time(positions, origin, destination):
    return h_distance(positions, origin, destination) / MAX_SPEED
```

#### Task 6 - A*

```python
def a_star(graph, origin, destination):
    priority_queue = [(h_time(positions, origin, destination), 0, origin, [origin])]
    visited = set()


    while priority_queue:
        f, cost, current, path = heapq.heappop(priority_queue)


        if current in visited:
            continue
        
        if current == destination:
            return path, cost


        visited.add(current)

        for neighbour in graph[current]:
            if neighbour in visited:
                continue

            new_cost = cost + graph[current][neighbour]["time_min"]
            new_f = new_cost + h_time(positions, neighbour, destination)
            heapq.heappush(priority_queue, (new_f, new_cost, neighbour, path + [neighbour]))


    return None





a_star(graph, "budapest", "rome")
```
